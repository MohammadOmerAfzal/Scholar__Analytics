# Statistical Analysis Part I

Introduction
Statistical analysis is the process of collecting, organizing, analyzing, and interpreting data to discover patterns, relationships, and meaningful insights that support decision-making.
It’s about turning raw numbers into useful information. Instead of just having data, you use methods to understand what the data is telling you.
Statistical analysis is built on the idea of summarizing data and making conclusions using concepts from Statistics.
In the following analysis, we perform a squential statistical analysis that will help the basic to advanced learners with the help of Python. 
We have used CARS dataset to analyse and obtain the results. 

```python
#importing two libraries pandas and numpy for statistical analysis  
import pandas as pd
import numpy as np
```

```python
# Reading dataset from local disk where file is saved. You should use 'r' before quotation marks of path so that slashes can be reversed. This is requirement of path
# File can be read in any format including csv, excel, SAS, Stata. We just need to give function of i.e. read_csv or read_sas etc
# If path does not recognize. upload your file in Jupypter and then read it by looking where your Jypyter is installed.
cars= pd.read_excel(r"C:\Users\Admin\Downloads\CARS.xlsx")
```

```python
# .head() is a function to read first five observations. You can increase that number by inserting values in bracket i.e. (10)
cars.head()
```

```python
# The function .head() is to read last five observations. You can increase that number by inserting values in bracket i.e. (10)
cars.tail()
```

```python
# We want to know the number of observations and colums.
cars.shape
```

```python
# .info() function provides basic information about variables and their types including fload, integers and object. This is crucial in knwoing types of variables in your dataset
# It also tells number of obvservations in each variable and missing, nonmissing values.
cars.info()
```

```python
# .columns tells about total variables and name of variables in actual which are written in dataset. 
# Important thing is they are shown in quotation marks and if you use them in further analysis, they can be copied. It will help avoid small or big cap problem
cars.columns
```

```python
# If you want to view only on variable then you can use the following code by connecting dataset name with variable name
cars.Make
```

```python
# if you want to analyze using any statistic for specific variable i.e. mean, median, sum etc. 
cars['EngineSize'].mean()
```

```python
# if you want to mean/median/min/max of each numeric variable 
cars.mean(numeric_only=True)
```

```python
# If you want mean/median/min/max of each row rather columns
cars.mean(axis = 1,numeric_only=True)
```

```python
# We want descriptive statistics of all numeric variables. 
# Further output has more decimal places. So, we can use .round() function to reduce decimal places. We can add .round(2) upto two decimal places
cars.describe().round()
```

```python
# We can also transpose out output as required by research thesis/papers. We can add only .T at the end of code
cars.describe().round().T
```

```python
# Previous code only provide 25%, 50%, 75% percentiles. If we want to extend our percentiles, we can customize as per our requirements
cars.describe(include=[np.number], percentiles=[.01, .05, .10, .25, .5, .75, .9, .95, .99]).T
```

```python
#Finally, we can also get our categorical values to observe each variable with maximum categories along with Mode of each category.
cars.describe(exclude=[np.number])
```
